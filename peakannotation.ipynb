```r
# ============================================================
# Sheep Oar_v4.0 ChIPseeker peak annotation
# ============================================================

# ============================================================
# 0. 加载 R 包
# ============================================================

library(ChIPseeker)
library(GenomicFeatures)
library(GenomicRanges)
library(rtracklayer)
library(dplyr)
library(ggplot2)


# ============================================================
# 2. 四个 peak 文件
# ============================================================

peak_files <- c(
    "HTY-ovary-H3K27ac.merged.bed",
    "HTY-ovary-H3K4me3.merged.bed",
    "HY-ovary-H3K27ac.merged.bed",
    "HY-ovary-H3K4me3.merged.bed"
)

sample_names <- c(
    "HTY_ovary_H3K27ac",
    "HTY_ovary_H3K4me3",
    "HY_ovary_H3K27ac",
    "HY_ovary_H3K4me3"
)


# ============================================================
# 3. 检查 peak 文件是否存在
# ============================================================

missing_files <- peak_files[!file.exists(peak_files)]

if (length(missing_files) > 0) {
    stop(
        paste(
            "以下文件不存在:",
            paste(missing_files, collapse = "\n")
        )
    )
}


# ============================================================
# 4. 创建输出目录
# ============================================================

dir.create(
    "ChIPseeker_annotation",
    showWarnings = FALSE
)

dir.create(
    "ChIPseeker_annotation/results",
    showWarnings = FALSE
)

dir.create(
    "ChIPseeker_annotation/plots",
    showWarnings = FALSE
)


# ============================================================
# 5. 构建 Oar_v4.0 TxDb
# ============================================================

txdb_file <- "/home/hnadmin/data/yangshuju/result/TxDb.Ovis_aries.Oar_v4.0.sqlite"

if (file.exists(txdb_file)) {

    message("发现已有 TxDb，直接读取...")

    txdb <- loadDb(txdb_file)

} else {

    message("没有发现 TxDb，开始从 GTF 构建...")

    txdb <- makeTxDbFromGFF(
        gtf_file,
        format = "gtf"
    )

    saveDb(
        txdb,
        file = txdb_file
    )

}

print(txdb)


# ============================================================
# 6. 查看 TxDb chromosome
# ============================================================

tx_chromosomes <- seqlevels(txdb)

cat("\nTxDb chromosomes:\n")
print(tx_chromosomes)


# ============================================================
# 7. 读取四个 BED
# ============================================================

peak_gr_list <- lapply(
    peak_files,
    function(x) {

        message("Reading: ", x)

        gr <- import(x)

        return(gr)
    }
)

names(peak_gr_list) <- sample_names


# ============================================================
# 8. 检查 BED chromosome
# ============================================================

for (i in seq_along(peak_gr_list)) {

    cat(
        "\n",
        sample_names[i],
        "chromosomes:\n"
    )

    print(
        head(
            seqlevels(peak_gr_list[[i]]),
            20
        )
    )
}


# ============================================================
# 9. 检查 BED 和 GTF chromosome 命名是否一致
# ============================================================

for (i in seq_along(peak_gr_list)) {

    peak_chr <- seqlevels(peak_gr_list[[i]])

    overlap_chr <- intersect(
        peak_chr,
        tx_chromosomes
    )

    cat(
        "\n",
        sample_names[i],
        "\n",
        "Peak chromosomes: ",
        length(peak_chr),
        "\n",
        "Matched chromosomes: ",
        length(overlap_chr),
        "\n"
    )

    if (length(overlap_chr) == 0) {

        warning(
            sample_names[i],
            ": BED 和 GTF 的 chromosome 命名完全不一致!"
        )

    }
}


# ============================================================
# 10. 如果 chromosome 命名一致，进行 ChIPseeker 注释
# ============================================================

peak_annotation <- lapply(
    seq_along(peak_gr_list),
    function(i) {

        message(
            "\nAnnotating: ",
            sample_names[i]
        )

        annotatePeak(
            peak_gr_list[[i]],
            TxDb = txdb,

            # promoter 定义
            tssRegion = c(-3000, 3000),

            # 不限制 chromosome
            verbose = FALSE
        )
    }
)

names(peak_annotation) <- sample_names


# ============================================================
# 11. 保存 ChIPseeker 对象
# ============================================================

saveRDS(
    peak_annotation,
    file = "ChIPseeker_annotation/peak_annotation.rds"
)


# ============================================================
# 12. 导出每一个 peak 的完整 annotation
# ============================================================

for (i in seq_along(peak_annotation)) {

    anno_df <- as.data.frame(
        peak_annotation[[i]]
    )

    output_file <- paste0(
        "ChIPseeker_annotation/results/",
        sample_names[i],
        ".ChIPseeker.annotation.tsv"
    )

    write.table(
        anno_df,
        file = output_file,
        sep = "\t",
        quote = FALSE,
        row.names = FALSE
    )

    message(
        "Saved: ",
        output_file
    )
}


# ============================================================
# 13. 查看第一个结果
# ============================================================

head(
    as.data.frame(
        peak_annotation[[1]]
    )
)


# ============================================================
# 14. 提取 genomic annotation
# ============================================================

annotation_df <- lapply(
    seq_along(peak_annotation),
    function(i) {

        df <- as.data.frame(
            peak_annotation[[i]]
        )

        df %>%
            count(annotation) %>%
            mutate(
                sample = sample_names[i],
                percentage = n / sum(n) * 100
            )
    }
) %>%
    bind_rows()


# ============================================================
# 15. 保存 annotation summary
# ============================================================

write.table(
    annotation_df,
    file = "ChIPseeker_annotation/results/genomic_annotation_summary.tsv",
    sep = "\t",
    quote = FALSE,
    row.names = FALSE
)


# ============================================================
# 16. 绘制 genomic annotation stacked barplot
# ============================================================

p_annotation <- ggplot(
    annotation_df,
    aes(
        x = sample,
        y = percentage,
        fill = annotation
    )
) +

    geom_bar(
        stat = "identity",
        position = "stack"
    ) +

    theme_classic() +

    labs(
        x = NULL,
        y = "Peak percentage (%)",
        fill = "Genomic annotation"
    ) +

    theme(
        axis.text.x = element_text(
            angle = 45,
            hjust = 1
        ),
        axis.text = element_text(
            size = 11
        ),
        axis.title = element_text(
            size = 12
        ),
        legend.title = element_text(
            size = 11
        ),
        legend.text = element_text(
            size = 10
        )
    )

print(p_annotation)


# ============================================================
# 17. 保存 annotation 图
# ============================================================

ggsave(
    "ChIPseeker_annotation/plots/genomic_annotation.pdf",
    p_annotation,
    width = 10,
    height = 7
)

ggsave(
    "ChIPseeker_annotation/plots/genomic_annotation.png",
    p_annotation,
    width = 10,
    height = 7,
    dpi = 300
)


# ============================================================
# 18. 单独绘制四个 annotation pie plot
# ============================================================

for (i in seq_along(peak_annotation)) {

    p <- plotAnnoPie(
        peak_annotation[[i]]
    )

    pdf(
        paste0(
            "ChIPseeker_annotation/plots/",
            sample_names[i],
            "_annotation_pie.pdf"
        ),
        width = 7,
        height = 7
    )

    print(p)

    dev.off()
}


# ============================================================
# 19. 绘制四个 peak 到 TSS 的距离分布
# ============================================================

for (i in seq_along(peak_annotation)) {

    p <- plotDistToTSS(
        peak_annotation[[i]]
    )

    pdf(
        paste0(
            "ChIPseeker_annotation/plots/",
            sample_names[i],
            "_distance_to_TSS.pdf"
        ),
        width = 8,
        height = 6
    )

    print(p)

    dev.off()
}


# ============================================================
# 20. 提取 peak → gene 信息
# ============================================================

gene_annotation_df <- lapply(
    seq_along(peak_annotation),
    function(i) {

        df <- as.data.frame(
            peak_annotation[[i]]
        )

        # 根据 ChIPseeker 输出自动保留存在的列
        keep_cols <- intersect(
            c(
                "seqnames",
                "start",
                "end",
                "width",
                "strand",
                "annotation",
                "geneChr",
                "geneStart",
                "geneEnd",
                "geneLength",
                "geneStrand",
                "geneId",
                "distanceToTSS"
            ),
            colnames(df)
        )

        df <- df[, keep_cols, drop = FALSE]

        df$sample <- sample_names[i]

        df
    }
) %>%
    bind_rows()


# ============================================================
# 21. 保存 peak-gene annotation
# ============================================================

write.table(
    gene_annotation_df,
    file = "ChIPseeker_annotation/results/peak_gene_annotation.tsv",
    sep = "\t",
    quote = FALSE,
    row.names = FALSE
)


# ============================================================
# 22. 统计四个样本的 peak 数量
# ============================================================

peak_number_summary <- data.frame(
    sample = sample_names,
    peak_number = sapply(
        peak_gr_list,
        length
    )
)

print(
    peak_number_summary
)

write.table(
    peak_number_summary,
    file = "ChIPseeker_annotation/results/peak_number_summary.tsv",
    sep = "\t",
    quote = FALSE,
    row.names = FALSE
)


# ============================================================
# 23. 统计 annotation 数量
# ============================================================

annotation_count <- annotation_df %>%
    select(
        sample,
        annotation,
        n
    )

print(annotation_count)

write.table(
    annotation_count,
    file = "ChIPseeker_annotation/results/annotation_count.tsv",
    sep = "\t",
    quote = FALSE,
    row.names = FALSE
)


# ============================================================
# 24. 统计 promoter peak
# ============================================================

promoter_summary <- annotation_df %>%
    filter(
        grepl(
            "Promoter",
            annotation,
            ignore.case = TRUE
        )
    ) %>%
    select(
        sample,
        n,
        percentage
    )

print(promoter_summary)


# ============================================================
# 25. 完成
# ============================================================

cat("\n")
cat("=============================================\n")
cat("ChIPseeker annotation finished!\n")
cat("=============================================\n")

cat(
    "\nResults:\n",
    "ChIPseeker_annotation/results/\n"
)

cat(
    "\nPlots:\n",
    "ChIPseeker_annotation/plots/\n"
)
```

```r
library(ChIPseeker)
library(AnnotationDbi)
library(clusterProfiler)
library(GenomicFeatures)
library(dplyr)
library(stringr)
library(readr)
library(ggplot2)
# ============================================================
# 0. 数据库
# ============================================================

txdb <- loadDb("/home/hnadmin/data/yangshuju/result/TxDb.Ovis_aries.Oar_v4.0.sqlite")
orgdb <- loadDb("/home/hnadmin/data/yangshuju/result/org.Ovis_aries.eg.sqlite")


# ============================================================
# 1. 读取 peak annotation
# ============================================================
# 如果当前 R 环境中已经有 peak_annotation，可以跳过这一部分
#
# 假设之前保存的是：
# peak_annotation.rds

peak_annotation <- readRDS(
    "/home/hnadmin/data/yangshuju/result/ovary/peak/ChIPseeker_annotation/peak_annotation.rds"
)

sample_names <- c(
    "HTY_H3K27ac",
    "HTY_H3K4me3",
    "HY_H3K27ac",
    "HY_H3K4me3"
)

names(peak_annotation) <- sample_names


# ============================================================
# 2. 创建输出目录
# ============================================================

dir.create(
    "GO_analysis",
    showWarnings = FALSE
)

dir.create(
    "GO_analysis/01_independent",
    showWarnings = FALSE,
    recursive = TRUE
)

dir.create(
    "GO_analysis/02_HY_vs_HTY",
    showWarnings = FALSE,
    recursive = TRUE
)

dir.create(
    "GO_analysis/03_specific_genes",
    showWarnings = FALSE,
    recursive = TRUE
)

dir.create(
    "GO_analysis/plots",
    showWarnings = FALSE,
    recursive = TRUE
)


# ============================================================
# 3. 提取 gene ID
# ============================================================

extract_genes <- function(annotation_obj) {
    
    df <- as.data.frame(annotation_obj)
    
    genes <- df$geneId
    
    genes <- genes[
        !is.na(genes) &
        genes != "" &
        genes != "NA"
    ]
    
    unique(as.character(genes))
}


HTY_H3K27ac_genes <- extract_genes(
    peak_annotation[["HTY_H3K27ac"]]
)

HTY_H3K4me3_genes <- extract_genes(
    peak_annotation[["HTY_H3K4me3"]]
)

HY_H3K27ac_genes <- extract_genes(
    peak_annotation[["HY_H3K27ac"]]
)

HY_H3K4me3_genes <- extract_genes(
    peak_annotation[["HY_H3K4me3"]]
)


# ============================================================
# 4. 检查 gene ID
# ============================================================

cat("\n============================\n")
cat("Gene numbers\n")
cat("============================\n")

cat("HTY H3K27ac:", length(HTY_H3K27ac_genes), "\n")
cat("HTY H3K4me3:", length(HTY_H3K4me3_genes), "\n")
cat("HY H3K27ac :", length(HY_H3K27ac_genes), "\n")
cat("HY H3K4me3 :", length(HY_H3K4me3_genes), "\n")


# ============================================================
# 5. OrgDb 中所有 Entrez gene
# ============================================================

all_genes <- keys(
    orgdb,
    keytype = "SYMBOL"
)

all_genes <- unique(
    all_genes[
        !is.na(all_genes) &
        all_genes != ""
    ]
)

cat(
    "OrgDb genes:",
    length(all_genes),
    "\n"
)


# ============================================================
# 6. 检查 gene ID 能否匹配 OrgDb
# ============================================================

check_genes <- function(genes, name) {
    
    matched <- intersect(
        genes,
        all_genes
    )
    
    cat(
        name,
        ":",
        length(genes),
        "input ->",
        length(matched),
        "matched\n"
    )
    
    return(matched)
}


HTY_H3K27ac_genes <- check_genes(
    HTY_H3K27ac_genes,
    "HTY H3K27ac"
)

HTY_H3K4me3_genes <- check_genes(
    HTY_H3K4me3_genes,
    "HTY H3K4me3"
)

HY_H3K27ac_genes <- check_genes(
    HY_H3K27ac_genes,
    "HY H3K27ac"
)

HY_H3K4me3_genes <- check_genes(
    HY_H3K4me3_genes,
    "HY H3K4me3"
)


# ============================================================
# ============================================================
# 方案一
# HTY / HY 分别进行 GO enrichment
# ============================================================
# ============================================================

run_GO <- function(
    genes,
    universe,
    prefix
) {
    
    result_list <- list()
    
    for (ont in c("BP")) {
        
        cat(
            "\nRunning:",
            prefix,
            ont,
            "\n"
        )
        
        ego <- enrichGO(
            gene = genes,
            universe = universe,
            OrgDb = orgdb,
            keyType = "SYMBOL",
            ont = ont,
            pAdjustMethod = "BH",
            pvalueCutoff = 0.05,
            qvalueCutoff = 0.2,
            readable = TRUE
        )
        
        result_list[[ont]] <- ego
        
        if (!is.null(ego)) {
            
            result_df <- as.data.frame(ego)
            
            if (nrow(result_df) > 0) {
                
                write.csv(
                    result_df,
                    paste0(
                        prefix,
                        "_",
                        ont,
                        ".csv"
                    ),
                    row.names = FALSE
                )
                
                saveRDS(
                    ego,
                    paste0(
                        prefix,
                        "_",
                        ont,
                        ".rds"
                    )
                )
            }
        }
    }
    
    return(result_list)
}


# ============================================================
# 方案一：四组
# ============================================================

GO_01_HTY_H3K27ac <- run_GO(
    HTY_H3K27ac_genes,
    all_genes,
    "GO_analysis/01_independent/HTY_H3K27ac"
)

GO_01_HTY_H3K4me3 <- run_GO(
    HTY_H3K4me3_genes,
    all_genes,
    "GO_analysis/01_independent/HTY_H3K4me3"
)

GO_01_HY_H3K27ac <- run_GO(
    HY_H3K27ac_genes,
    all_genes,
    "GO_analysis/01_independent/HY_H3K27ac"
)

GO_01_HY_H3K4me3 <- run_GO(
    HY_H3K4me3_genes,
    all_genes,
    "GO_analysis/01_independent/HY_H3K4me3"
)


# ============================================================
# ============================================================
# 方案二
# HY-specific = HY - HTY
# HTY gene set作为background
# ============================================================
# ============================================================

run_HY_vs_HTY <- function(
    HY_genes,
    HTY_genes,
    prefix
) {
    
    # HY-specific
    HY_specific <- setdiff(
        HY_genes,
        HTY_genes
    )
    
    # 只保留 OrgDb 中存在的 gene
    HY_specific <- intersect(
        HY_specific,
        all_genes
    )
    
    HTY_background <- intersect(
        HTY_genes,
        all_genes
    )
    
    cat("\n============================\n")
    cat("HY vs HTY\n")
    cat("============================\n")
    
    cat(
        "HTY background:",
        length(HTY_background),
        "\n"
    )
    
    cat(
        "HY-specific:",
        length(HY_specific),
        "\n"
    )
    
    # 保存 gene list
    writeLines(
        HY_specific,
        paste0(
            prefix,
            "_HY_specific_genes.txt"
        )
    )
    
    writeLines(
        HTY_background,
        paste0(
            prefix,
            "_HTY_background_genes.txt"
        )
    )
    
    result_list <- list()
    
    for (ont in c("BP")) {
        
        ego <- enrichGO(
            gene = HY_specific,
            universe = HTY_background,
            OrgDb = orgdb,
            keyType = "SYMBOL",
            ont = ont,
            pAdjustMethod = "BH",
            pvalueCutoff = 0.05,
            qvalueCutoff = 0.2,
            readable = TRUE
        )
        
        result_list[[ont]] <- ego
        
        if (!is.null(ego)) {
            
            result_df <- as.data.frame(ego)
            
            if (nrow(result_df) > 0) {
                
                write.csv(
                    result_df,
                    paste0(
                        prefix,
                        "_",
                        ont,
                        ".csv"
                    ),
                    row.names = FALSE
                )
                
                saveRDS(
                    ego,
                    paste0(
                        prefix,
                        "_",
                        ont,
                        ".rds"
                    )
                )
            }
        }
    }
    
    return(result_list)
}


# ============================================================
# H3K27ac
# ============================================================

GO_02_H3K27ac <- run_HY_vs_HTY(
    HY_H3K27ac_genes,
    HTY_H3K27ac_genes,
    "GO_analysis/02_HY_vs_HTY/H3K27ac"
)


# ============================================================
# H3K4me3
# ============================================================

GO_02_H3K4me3 <- run_HY_vs_HTY(
    HY_H3K4me3_genes,
    HTY_H3K4me3_genes,
    "GO_analysis/02_HY_vs_HTY/H3K4me3"
)


# ============================================================
# ============================================================
# 方案三
# HY-specific vs HTY-specific
# ============================================================
# ============================================================

run_specific_GO <- function(
    HY_genes,
    HTY_genes,
    prefix
) {
    
    HY_specific <- setdiff(
        HY_genes,
        HTY_genes
    )
    
    HTY_specific <- setdiff(
        HTY_genes,
        HY_genes
    )
    
    HY_specific <- intersect(
        HY_specific,
        all_genes
    )
    
    HTY_specific <- intersect(
        HTY_specific,
        all_genes
    )
    
    cat("\n============================\n")
    cat(prefix, "\n")
    cat("============================\n")
    
    cat(
        "HY-specific:",
        length(HY_specific),
        "\n"
    )
    
    cat(
        "HTY-specific:",
        length(HTY_specific),
        "\n"
    )
    
    # 保存 gene list
    writeLines(
        HY_specific,
        paste0(
            prefix,
            "_HY_specific_genes.txt"
        )
    )
    
    writeLines(
        HTY_specific,
        paste0(
            prefix,
            "_HTY_specific_genes.txt"
        )
    )
    
    result <- list()
    
    # ----------------------------
    # HY-specific
    # ----------------------------
    
    result$HY_specific <- run_GO(
        HY_specific,
        all_genes,
        paste0(
            prefix,
            "_HY_specific"
        )
    )
    
    # ----------------------------
    # HTY-specific
    # ----------------------------
    
    result$HTY_specific <- run_GO(
        HTY_specific,
        all_genes,
        paste0(
            prefix,
            "_HTY_specific"
        )
    )
    
    return(result)
}


GO_03_H3K27ac <- run_specific_GO(
    HY_H3K27ac_genes,
    HTY_H3K27ac_genes,
    "GO_analysis/03_specific_genes/H3K27ac"
)


GO_03_H3K4me3 <- run_specific_GO(
    HY_H3K4me3_genes,
    HTY_H3K4me3_genes,
    "GO_analysis/03_specific_genes/H3K4me3"
)


# ============================================================
# ============================================================
# 7. GO 绘图函数
# ============================================================
# ============================================================

plot_GO <- function(
    ego,
    filename,
    title,
    show_n = 15
) {
    
    if (is.null(ego)) {
        return(NULL)
    }
    
    df <- as.data.frame(ego)
    
    if (nrow(df) == 0) {
        return(NULL)
    }
    
    df <- df %>%
        arrange(p.adjust) %>%
        head(show_n)
    
    # 保证最显著的 term 在最上面
    df$Description <- factor(
        df$Description,
        levels = rev(df$Description)
    )
    
    p <- ggplot(
        df,
        aes(
            x = Count,
            y = Description,
            size = Count,
            color = p.adjust
        )
    ) +
        geom_point() +
        scale_color_continuous(
            low = "red",
            high = "blue",
            trans = "reverse"
        ) +
        labs(
            title = title,
            x = "Gene count",
            y = NULL,
            color = "Adjusted P",
            size = "Gene count"
        ) +
        theme_classic() +
        theme(
            plot.title = element_text(
                hjust = 0.5,
                face = "bold",
                size = 14
            ),
            axis.text.y = element_text(
                size = 9
            ),
            legend.title = element_text(
                size = 10
            ),
            legend.text = element_text(
                size = 9
            )
        )
    
    ggsave(
        filename = filename,
        plot = p,
        width = 9,
        height = 7
    )
    
    print(p)
}


# ============================================================
# ============================================================
# 8. 绘制方案一
# ============================================================
# ============================================================

for (group in names(GO_01_HTY_H3K27ac)) {
    
    plot_GO(
        GO_01_HTY_H3K27ac[[group]],
        paste0(
            "GO_analysis/plots/01_HTY_H3K27ac_",
            group,
            ".pdf"
        ),
        paste(
            "HTY H3K27ac - GO",
            group
        )
    )
}


for (group in names(GO_01_HTY_H3K4me3)) {
    
    plot_GO(
        GO_01_HTY_H3K4me3[[group]],
        paste0(
            "GO_analysis/plots/01_HTY_H3K4me3_",
            group,
            ".pdf"
        ),
        paste(
            "HTY H3K4me3 - GO",
            group
        )
    )
}


for (group in names(GO_01_HY_H3K27ac)) {
    
    plot_GO(
        GO_01_HY_H3K27ac[[group]],
        paste0(
            "GO_analysis/plots/01_HY_H3K27ac_",
            group,
            ".pdf"
        ),
        paste(
            "HY H3K27ac - GO",
            group
        )
    )
}


for (group in names(GO_01_HY_H3K4me3)) {
    
    plot_GO(
        GO_01_HY_H3K4me3[[group]],
        paste0(
            "GO_analysis/plots/01_HY_H3K4me3_",
            group,
            ".pdf"
        ),
        paste(
            "HY H3K4me3 - GO",
            group
        )
    )
}


# ============================================================
# ============================================================
# 9. 绘制方案二
# ============================================================
# ============================================================

for (ont in c("BP")) {
    
    plot_GO(
        GO_02_H3K27ac[[ont]],
        paste0(
            "GO_analysis/plots/02_H3K27ac_HY_vs_HTY_",
            ont,
            ".pdf"
        ),
        paste(
            "H3K27ac HY-specific vs HTY",
            ont
        )
    )
    
    plot_GO(
        GO_02_H3K4me3[[ont]],
        paste0(
            "GO_analysis/plots/02_H3K4me3_HY_vs_HTY_",
            ont,
            ".pdf"
        ),
        paste(
            "H3K4me3 HY-specific vs HTY",
            ont
        )
    )
}


# ============================================================
# ============================================================
# 10. 绘制方案三
# ============================================================
# ============================================================

for (mark in c("H3K27ac", "H3K4me3")) {
    
    go_obj <- if (
        mark == "H3K27ac"
    ) {
        GO_03_H3K27ac
    } else {
        GO_03_H3K4me3
    }
    
    for (group in c(
        "HY_specific",
        "HTY_specific"
    )) {
        
        for (ont in c(
            "BP",
            "MF",
            "CC"
        )) {
            
            plot_GO(
                go_obj[[group]][[ont]],
                paste0(
                    "GO_analysis/plots/03_",
                    mark,
                    "_",
                    group,
                    "_",
                    ont,
                    ".pdf"
                ),
                paste(
                    mark,
                    group,
                    ont
                )
            )
        }
    }
}


# ============================================================
# 11. 保存所有 GO 对象
# ============================================================

saveRDS(
    list(
        independent = list(
            HTY_H3K27ac = GO_01_HTY_H3K27ac,
            HTY_H3K4me3 = GO_01_HTY_H3K4me3,
            HY_H3K27ac = GO_01_HY_H3K27ac,
            HY_H3K4me3 = GO_01_HY_H3K4me3
        ),
        HY_vs_HTY = list(
            H3K27ac = GO_02_H3K27ac,
            H3K4me3 = GO_02_H3K4me3
        ),
        specific = list(
            H3K27ac = GO_03_H3K27ac,
            H3K4me3 = GO_03_H3K4me3
        )
    ),
    "GO_analysis/all_GO_results.rds"
)


cat("\n========================================\n")
cat("ALL GO ANALYSES FINISHED\n")
cat("========================================\n")
```

```r
run_GO_BP <- function(
    genes,
    universe_genes,
    OrgDb,
    output_prefix
) {
    
    if (length(genes) < 5) {
        cat(output_prefix, ": overlap gene < 5, skip GO\n")
        return(NULL)
    }
    
    ego <- enrichGO(
        gene          = genes,
        universe      = universe_genes,
        OrgDb         = OrgDb,
        keyType       = "SYMBOL",
        ont           = "BP",
        pAdjustMethod = "BH",
        pvalueCutoff  = 0.05,
        #qvalueCutoff  = 0.1,
        readable      = TRUE
    )
    
    # 保存结果
    if (!is.null(ego) && nrow(as.data.frame(ego)) > 0) {
        
        write.csv(
            as.data.frame(ego),
            paste0(output_prefix, "_GO_BP.csv"),
            row.names = FALSE
        )
        
        # Dotplot
        p <- dotplot(
            ego,
            showCategory = 20
        ) +
            ggtitle(output_prefix) +
            theme_classic() +
            theme(
                plot.title = element_text(hjust = 0.5)
            )
        
        ggsave(
            paste0(output_prefix, "_GO_BP.pdf"),
            p,
            width = 8,
            height = 6
        )
    }
    
    return(ego)
}
```
